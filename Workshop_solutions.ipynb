IMPORT REQUIRED LIBRARIES

```python
import numpy as np
import matplotlib.pyplot as plt
```

Genereate Random data

```python
np.random.seed(42)

income_growth = np.random.normal(loc=5, scale=2, size=1000)

consumer_pref = np.random.uniform(low=0, high=10, size=1000)

unemployment_duration = np.random.exponential(scale=2, size=1000)

loan_repayment = np.random.beta(a=2, b=5, size=1000)
```

NORMAL DISTRIBUTION

```python
plt.figure(figsize=(6,4))
plt.hist(income_growth, bins=30, color='blue')
plt.title('Distribution of Household Income Growth Rates')
plt.xlabel('Growth Rates(%)')
plt.ylabel('Frequency')
plt.grid(True)
plt.show()
```

UNIFORM DISTRIBUTION

```python
plt.figure(figsize=(6,4))
plt.hist(consumer_pref, bins=30, color='green')
plt.title('Distribution of Consumer Preferences')
plt.xlabel('Preferences score')
plt.ylabel('Frequency')
plt.grid(True)
```

Exponential distribution

```python
plt.figure(figsize=(6,4))
plt.hist(unemployment_duration, bins=30, color='orange')
plt.title('Distribution of Unemployment Duration')
plt.xlabel('Duration(months)')
plt.ylabel('Frequency')
plt.grid(True)
#
```

BETA DISTRIBUTION

```python
plt.figure(figsize=(6,4))
plt.hist(loan_repayment, bins=30, color='red')
plt.title('Distribution of Loan Repayment Probability')
plt.xlabel('Probability')
plt.ylabel('Frequency')
plt.grid(True)
```

COMBINED PLOTS

```python
fig, ax = plt.subplots(2, 2, figsize=(12, 8))

ax[0,0].hist(income_growth, bins=30, color='blue')
ax[0,0].set_title('Normal Distribution')

ax[0,1].hist(consumer_pref, bins=30, color='green')
ax[0,1].set_title('Uniform Distribution')

ax[1,0].hist(unemployment_duration, bins=30, color='orange')
ax[1,0].set_title('Exponential Distribution')

ax[1,1].hist(loan_repayment, bins=30, color='red')
ax[1,1].set_title('Beta Distribution')

fig.suptitle(
    'Economic Variables Generated from Different Probability Distributions',
fontsize = 14, fontweight = 'bold'
)

plt.tight_layout()
plt.show()
```
